```python
# switch half of the Ni into Co and half Ni into Zr
from ase.io import vasp
from ase.io import lammpsdata
import random
import numpy as np

NiTi = vasp.read_vasp("./POSCAR")


def random_index(start, end, num):
    random.seed(12345678)
    index = random.sample(range(start, end), num)
    index = np.array(index)
    index.sort()
    index = index.tolist()
    return index


index_NiCo = random_index(0, 22848, 11424)
for i in index_NiCo:
    NiTi[i].symbol = "Co"

index_NiZr = random_index(23040, 45696, 11424)
for i in index_NiZr:
    NiTi[i].symbol = "Zr"

print(NiTi.get_chemical_formula())
# vasp.write_vasp("coursework_practice/class/481048.pos", NiTi)
lammpsdata.write_lammps_data(
    "./NiCoTiZrHf_45696.lmp", NiTi, specorder=["Ni", "Co", "Ti", "Zr"], masses=True
)
```
