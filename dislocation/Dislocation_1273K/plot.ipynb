```python
import matplotlib.pyplot as plt
import numpy as np

# Data from the image
labels = ['b100_p100', 'b100_p110', 'b110_p100', 'b110_p110']
edge_values = [1.354016338
,1.010099782
,8.506905197
,3.026030806
]
screw_values = [0.234646751
,0.098100332
,2.902111855
,1.29155243
]

# X-axis locations for the groups
x = np.arange(len(labels))

# Width of the bars
width = 0.35

# Plotting the bars
fig, ax = plt.subplots()
bars1 = ax.bar(x - width/2, edge_values, width, label='Edge')
bars2 = ax.bar(x + width/2, screw_values, width, label='Screw')

# Adding labels, title, and custom x-axis tick labels
# ax.set_xlabel('b_p_ Configurations')
ax.set_ylabel('formation energy (eV/Å)')
ax.set_title('Formation Energy in \n Different Burgers Vector and Slip Plane (1273K)')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()

# Display the plot
plt.xticks(rotation=45)
plt.show()
```
