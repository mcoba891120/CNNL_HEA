```python
import pandas as pd
import matplotlib.pyplot as plt

# 加载数据并忽略第一行
file_path = 'Record_npt.txt'
columns = ['Step', 'Temperature', 'Potential_Energy', 'Volume', 'Density', 'Pxx', 'Pyy', 'Pzz']

# 读取数据，跳过第一行注释
data = pd.read_csv(file_path, sep=' ', header=None, names=columns, skiprows=1)

# 计算平均密度
average_density = data['Density'].mean()

# 创建子图
fig, axs = plt.subplots(4, 2, figsize=(20, 20))

# 时间对温度作图
axs[0, 0].plot(data['Step'], data['Temperature'], marker='o', linestyle='-', color='b')
axs[0, 0].set_xlabel('Step')
axs[0, 0].set_ylabel('Temperature (K)')
axs[0, 0].set_title('Temperature vs Step')
axs[0, 0].grid(True)

# 时间对势能作图
axs[0, 1].plot(data['Step'], data['Potential_Energy'], marker='o', linestyle='-', color='r')
axs[0, 1].set_xlabel('Step')
axs[0, 1].set_ylabel('Potential Energy (eV)')
axs[0, 1].set_title('Potential Energy vs Step')
axs[0, 1].grid(True)

# 时间对体积作图
axs[1, 0].plot(data['Step'], data['Volume'], marker='o', linestyle='-', color='g')
axs[1, 0].set_xlabel('Step')
axs[1, 0].set_ylabel('Volume (A^3)')
axs[1, 0].set_title('Volume vs Step')
axs[1, 0].grid(True)

# 时间对密度作图并标示平均密度
axs[1, 1].plot(data['Step'], data['Density'], marker='o', linestyle='-', color='c', label='Density')
axs[1, 1].axhline(y=average_density, color='r', linestyle='--', label=f'Average Density = {average_density:.2f}')
axs[1, 1].set_xlabel('Step')
axs[1, 1].set_ylabel('Density (g/cm^3)')
axs[1, 1].set_title('Density vs Step')
axs[1, 1].legend()
axs[1, 1].grid(True)

# 时间对Pxx作图
axs[2, 0].plot(data['Step'], data['Pxx'], marker='o', linestyle='-', color='m')
axs[2, 0].set_xlabel('Step')
axs[2, 0].set_ylabel('Pxx (bar)')
axs[2, 0].set_title('Pxx vs Step')
axs[2, 0].grid(True)

# 时间对Pyy作图
axs[2, 1].plot(data['Step'], data['Pyy'], marker='o', linestyle='-', color='y')
axs[2, 1].set_xlabel('Step')
axs[2, 1].set_ylabel('Pyy (bar)')
axs[2, 1].set_title('Pyy vs Step')
axs[2, 1].grid(True)

# 时间对Pzz作图
axs[3, 0].plot(data['Step'], data['Pzz'], marker='o', linestyle='-', color='k')
axs[3, 0].set_xlabel('Step')
axs[3, 0].set_ylabel('Pzz (bar)')
axs[3, 0].set_title('Pzz vs Step')
axs[3, 0].grid(True)

# 调整布局
plt.tight_layout()

# 显示图表
plt.show()

# 打印平均密度
print(f'Average Density: {average_density}')

```
