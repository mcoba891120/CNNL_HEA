```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 读取数据
file_path = 'Record_thermal_exp.txt'
data = pd.read_csv(file_path, sep='\t')

# 显示前几行数据以确认读取正确
print(data.head())

# 计算斜率
slope, intercept = np.polyfit(data['Temp'], data['Strain'], 1)

# 绘制图表
plt.figure(figsize=(10, 6))
plt.plot(data['Temp'], data['Strain'], marker='o', linestyle='-', color='b', label='Data')
plt.plot(data['Temp'], slope * data['Temp'] + intercept, color='r', linestyle='--', label=f'Fit: Slope = {slope:.10f}')
plt.xlabel('Temperature (K)')
plt.ylabel('Strain')
plt.title('Thermal Expansion: Temperature vs Strain')
plt.legend()
plt.grid(True)
plt.show()

slope

```
