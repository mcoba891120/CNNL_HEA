```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
file_path = 'Record_npt.txt'
data = pd.read_csv(file_path, sep=' ', header=None, comment='#')

# 添加列名
data.columns = ['Timestep', 'Temperature', 'Potential_Energy', 'Kinetic_Energy', 'Volume', 'Pressure_xx', 'Pressure_yy', 'Pressure_zz']

# 显示前几行数据以确认读取正确
print(data.head())

# 选择初始体积
initial_volume = data['Volume'][0]

# 计算应变 (strain)
data['Strain'] = (data['Volume'] - initial_volume) / initial_volume

# 计算应力 (stress)，将压力从bar转换为Pa (1 bar = 1e5 Pa)
data['Stress'] = data['Pressure_xx'] * 1e5

# 绘制应力对应变的变化图
plt.figure(figsize=(10, 6))
plt.plot(data['Strain'], data['Stress'], marker='o', linestyle='-', color='b')
plt.xlabel('Strain')
plt.ylabel('Stress (Pa)')
plt.title('Stress vs Strain')
plt.grid(True)
plt.show()
```
