```python
import pandas as pd
import matplotlib.pyplot as plt

# 加载数据并忽略第一行
file_path = 'Record_npt.txt'
columns = ['Step', 'Temperature', 'Potential_Energy', 'Volume', 'Density', 'Pxx', 'Pyy', 'Pzz']

# 读取数据，跳过第一行注释
data = pd.read_csv(file_path, sep=' ', header=None, names=columns, skiprows=1)

# 计算平均密度
average_density = data['Density'].mean()

# 绘制密度随时间步变化图
plt.figure(figsize=(20, 6))
plt.plot(data['Step'], data['Density'], marker='o', linestyle='-', color='b', label='Density')
plt.axhline(y=average_density, color='r', linestyle='--', label=f'Average Density = {average_density:.2f}')
plt.xlabel('Step')
plt.ylabel('Density')
plt.title('Density vs Step')
plt.legend()
plt.grid(True)
plt.show()

print(f'Average Density: {average_density}')

```
