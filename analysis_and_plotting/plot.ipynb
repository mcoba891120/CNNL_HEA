```python
#use matplotlib to plot the data from txt file
import matplotlib.pyplot as plt
import numpy as np

record_npt = np.loadtxt('Record_npt.txt')

plt.figure(1)
plt.subplot(321)
plt.plot(record_npt[:,0], record_npt[:,1])
plt.title('Temperature')

plt.subplot(322)
plt.plot(record_npt[:,0], record_npt[:,2])
plt.title('Potential Energy')

plt.subplot(223)
plt.plot(record_npt[:,0], record_npt[:,3])
plt.title('Volume')

plt.subplot(224)
plt.plot(record_npt[:,0], record_npt[:,4])
plt.title('Density')
current_axis = plt.gca()
current_axis.set_yticks(np.arange(start=7.3, stop=7.6, step=0.02))
```

```python

record_thermal_exp = np.loadtxt('Record_thermal_exp.txt',skiprows=1)
plt.figure(2)
plt.plot(record_thermal_exp[:,0], record_thermal_exp[:,1])
slope = np.diff(record_thermal_exp[:,1])/np.diff(record_thermal_exp[:,0])
print('Average slope of the data is:', np.mean(slope))

plt.title('Thermal Expansion Coefficient')

plt.show()
```
