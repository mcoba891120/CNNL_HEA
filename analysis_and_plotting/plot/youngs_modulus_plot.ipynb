```python
import numpy as np
import matplotlib.pyplot as plt

line = var8_1500
# 取前两个点
strain_initial = line[:2, 0]
stress_initial = -line[:2, 1]

# 计算斜率 (杨氏模量) m = Δstress / Δstrain
slope = (stress_initial[1] - stress_initial[0]) / (strain_initial[1] - strain_initial[0])

# 创建延长线的应变范围
strain_extension = np.linspace(0, 0.05, 100)  # 可以根据你的需求调整最大应变值
stress_extension = slope * strain_extension  # 使用斜率计算延长线上的应力

# 指定的点
specified_points = [line[4,0],line[5,0],line[6,0]]  # 你可以根据需求指定点的位置
stress_values = [np.interp(p, line[:, 0], line[:, 1]) for p in specified_points]

# 绘图
plt.figure(figsize=(12, 6))

# 300K 数据
plt.plot(line[:, 0], line[:, 1], label='1500K', marker='o')

# 添加斜线表示弹性区域的延长线
plt.plot(strain_extension, stress_extension, color='r', linestyle='--', label=f'Elastic Line (slope = {slope:.2f})')

# 在指定点上绘制水平线并标注
for i, point in enumerate(specified_points):
    stress = stress_values[i]
    plt.axhline(y=stress, color='g', linestyle=':', xmax=point/0.05)
    plt.text(point + 0.009, stress, f'{stress:.2f}', color='g')

# 添加标签和标题
plt.xlabel('Strain')
plt.ylabel('Stress')
plt.title('Stress-Strain Curves for 300K with Elastic Region Extension')

# 显示图例
plt.legend()

# 显示网格
plt.grid(True)

# 调整布局
plt.tight_layout()

# 显示图表
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import linregress

# 文件路径列表
r300k = "compression/NiCoTiZrHf_100/var8_13500_300k/SS_curve_var8_13500_300k.txt"
r600k = "compression/NiCoTiZrHf_100/var8_13500_600k/SS_curve_var8_13500_600k.txt"
r900k = "compression/NiCoTiZrHf_100/var8_13500_900k/SS_curve_var8_13500_900k.txt"
r1200k = "compression/NiCoTiZrHf_100/var8_13500_1200k/SS_curve_var8_13500_1200k.txt"
r1500k = "compression/NiCoTiZrHf_100/var8_13500_1500k/SS_curve_var8_13500_1500k.txt"
temperatures = ['300K','600K', '900K','1200K', '1500K']
mc300_300k = "compression/NiCoTiZrHf_110/var8_12600_300k/SS_curve_var8_12600_300k.txt"
mc300_600k = "compression/NiCoTiZrHf_110/var8_12600_600k/SS_curve_var8_12600_600k.txt"
mc300_900k = "compression/NiCoTiZrHf_110/var8_12600_900k/SS_curve_var8_12600_900k.txt"
mc300_1200k = "compression/NiCoTiZrHf_110/var8_12600_1200k/SS_curve_var8_12600_1200k.txt"
mc300_1500k = "compression/NiCoTiZrHf_110/var8_12600_1500k/SS_curve_var8_12600_1500k.txt"
mc900_300k = "compression/NiCoTiZrHf_111/var8_12240_300k/SS_curve_var8_12240_300k.txt"
mc900_600k = "compression/NiCoTiZrHf_111/var8_12240_600k/SS_curve_var8_12240_600k.txt"
mc900_900k = "compression/NiCoTiZrHf_111/var8_12240_900k/SS_curve_var8_12240_900k.txt"
mc900_1200k = "compression/NiCoTiZrHf_111/var8_12240_1200k/SS_curve_var8_12240_1200k.txt"
mc900_1500k = "compression/NiCoTiZrHf_111/var8_12240_1500k/SS_curve_var8_12240_1500k.txt"
    
# 计算杨氏模量的函数
def calculate_youngs_modulus(file_path):
    # 读取数据文件，跳过标题行
    data = np.loadtxt(file_path, skiprows=1)
    # 仅使用前 20 个数据点
    strain = data[:20, 0]
    stress = data[:20, 1]
    # 计算线性回归的斜率，即杨氏模量
    slope, _, _, _, _ = linregress(strain, stress)
    return slope

# 计算每个数据集的杨氏模量
youngs_modulus = {
    '300K': calculate_youngs_modulus(r300k),
    '600K': calculate_youngs_modulus(r600k),
    '900K': calculate_youngs_modulus(r900k),
    '1200K': calculate_youngs_modulus(r1200k),
    '1500K': calculate_youngs_modulus(r1500k),
    '300K_mc300': calculate_youngs_modulus(mc300_300k),
    '600K_mc300': calculate_youngs_modulus(mc300_600k),
    '900K_mc300': calculate_youngs_modulus(mc300_900k),
    '1200K_mc300': calculate_youngs_modulus(mc300_1200k),
    '1500K_mc300': calculate_youngs_modulus(mc300_1500k),
    '300K_mc900': calculate_youngs_modulus(mc900_300k),
    '600K_mc900': calculate_youngs_modulus(mc900_600k),
    '900K_mc900': calculate_youngs_modulus(mc900_900k),
    '1200K_mc900': calculate_youngs_modulus(mc900_1200k),
    '1500K_mc900': calculate_youngs_modulus(mc900_1500k)
}

import matplotlib.pyplot as plt

# 温度列表
temperatures = ['300K','600K', '900K','1200K', '1500K']

# 杨氏模量数据列表
youngs_modulus_non_mc = [
    youngs_modulus['300K'],
    youngs_modulus['600K'],
    youngs_modulus['900K'],
    youngs_modulus['1200K'],
    youngs_modulus['1500K']
]

youngs_modulus_mc300 = [
    youngs_modulus['300K_mc300'],
    youngs_modulus['600K_mc300'],
    youngs_modulus['900K_mc300'],
    youngs_modulus['1200K_mc300'],
    youngs_modulus['1500K_mc300']
]

youngs_modulus_mc900 = [
    youngs_modulus['300K_mc900'],
    youngs_modulus['600K_mc900'],
    youngs_modulus['900K_mc900'],
    youngs_modulus['1200K_mc900'],
    youngs_modulus['1500K_mc900']
]

# 绘图
plt.figure(figsize=(10, 6))

# Non-MC 数据
plt.plot(temperatures, youngs_modulus_non_mc, marker='o', linestyle='-', color='b', label='100')

# 标注 Non-MC 数据的值
for i, temp in enumerate(temperatures):
    plt.text(temp, youngs_modulus_non_mc[i], f'{youngs_modulus_non_mc[i]:.2f}', ha='center', va='bottom', fontsize=12)

# MC300 数据
plt.plot(temperatures, youngs_modulus_mc300, marker='o', linestyle='-', color='r', label='110')

# 标注 MC300 数据的值
for i, temp in enumerate(temperatures):
    plt.text(temp, youngs_modulus_mc300[i], f'{youngs_modulus_mc300[i]:.2f}', ha='center', va='bottom', fontsize=12)

# MC900 数据
plt.plot(temperatures, youngs_modulus_mc900, marker='o', linestyle='-', color='g', label='111')

# 标注 MC900 数据的值
for i, temp in enumerate(temperatures):
    plt.text(temp, youngs_modulus_mc900[i], f'{youngs_modulus_mc900[i]:.2f}', ha='center', va='bottom', fontsize=12)

# 设置图表标题和标签
plt.title("Young's Modulus vs Temperature", fontsize=16)
plt.xlabel('Temperature (K)', fontsize=14)
plt.ylabel("Young's Modulus (GPa)", fontsize=14)

# 显示图例和网格
plt.legend()
plt.grid(True)

# 调整布局并显示图表
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data for edge and screw
categories = ['<100> {100}', '<100> {110}', '<110> {110}', '<111> {110}']
edge_values = [0.633120134512544, 1.24318092309623, 1.08040051638747, 2.41707711331917]
screw_values = [0.230507640577423, 0.464097768713136, 0.646941638136455, 1.3581592752886]

# Set the positions and width for the bars
x = np.arange(len(categories))
width = 0.35  # width of the bars

# Plotting the bars
fig, ax = plt.subplots()
bars1 = ax.bar(x - width/2, edge_values, width, label='Edge', color='b')
bars2 = ax.bar(x + width/2, screw_values, width, label='Screw', color='r')

# Add some text for labels, title, and axes ticks
ax.set_xlabel('slip system')
ax.set_ylabel('Formation Energy(eV/armstrong)')
ax.set_title('Formation energy of Edge and Screw dislocation under 300k')
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend()

# Add a function to label the bars with their values
def add_labels(bars):
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')

add_labels(bars1)
add_labels(bars2)

# Display the figure
plt.show()
```
