```python
import pandas as pd

# 讀取兩個 CSV 文件
file1 = "monte_carlo/var5_4000_900k/mc_folder/neighbor_analysis.csv"
file2 = "monte_carlo/var5_4000_900k/neighbor_analysis.csv"

df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

def percentage_to_float(df):
    for col in df.columns[1:]:
        df[col] = df[col].str.rstrip('%').astype('float') / 100
    return df

# 應用轉換
df1 = percentage_to_float(df1)
df2 = percentage_to_float(df2)

# 確保兩個表格有相同的結構，並刪除非數值的列
df1_numeric = df1.drop(columns=['Atom Type'])
df2_numeric = df2.drop(columns=['Atom Type'])

# 計算差值
difference = df1_numeric - df2_numeric

# 將差值轉換為百分比格式
difference_percentage = difference.applymap(lambda x: f"{x*100:.2f}%")

# 把 'Atom Type' 列加回來
difference_percentage.insert(0, 'Atom Type', df1['Atom Type'])

# 顯示結果
print(difference_percentage.to_string(index=False))

# 保存結果到新的 CSV 文件
difference_percentage.to_csv("neighbor_difference_percentage.csv", index=False)
```
