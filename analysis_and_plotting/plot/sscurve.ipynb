```python
from scipy.stats import linregress
import numpy as np
import matplotlib.pyplot as plt

def calculate_youngs_modulus(data):
    """Calculate Young's modulus using the first 20 data points of the stress-strain curve."""
    strain = data[:20, 0]
    stress = data[:20, 1]
    slope, _, _, _, _ = linregress(strain, stress)
    return slope

# Base file path pattern
base_path = "compression/NiCoTiZrHf_110/5quinary/v5_trial{trial}/{file_name}"

# Trial numbers
trials = [1, 2, 3, 4]

# Orientations
orientations = ["b100p110", "b111p110"]

# Temperatures
temperatures = [300, 600, 900]

# Colors for temperatures
temp_colors = {300: 'blue', 600: 'green', 900: 'red'}

# Line styles for orientations
orient_styles = {"b100p110": '-', "b111p110": '--'}

# Create one figure for each trial
for trial in trials:
    plt.figure(figsize=(12, 8))
    
    for temp in temperatures:
        for orient in orientations:
            # Construct the file name based on the pattern
            file_name = f"v5_trial{trial}_9792_{temp}k_{orient}/SS_curve_v5_trial{trial}_9792_{temp}k_{orient}.txt"
            full_path = base_path.format(trial=trial, file_name=file_name)
            
            try:
                # Load the data
                data = np.loadtxt(full_path, skiprows=1)
                
                # Calculate Young's modulus
                slope = calculate_youngs_modulus(data)
                
                # Plot the stress-strain curve
                label = f"{orient}, {temp}K (E: {slope:.2f})"
                plt.plot(data[:, 0], data[:, 1], 
                         label=label,
                         color=temp_colors[temp],
                         linestyle=orient_styles[orient],
                         linewidth=2)
                
            except FileNotFoundError:
                print(f"File not found: {full_path}")
            except Exception as e:
                print(f"Error processing {full_path}: {e}")
    
    # Set plot labels and title
    plt.xlabel('Strain', fontsize=14)
    plt.ylabel('Stress', fontsize=14)
    plt.title(f"Stress-Strain Curves for Trial {trial} at Different Temperatures and Orientations", fontsize=16)
    plt.legend(fontsize=12)
    plt.grid(True)
    
    # Add a text box with legend explanation
    plt.annotate(
        "Line styles:\n"
        "Solid: b100p110\n"
        "Dashed: b111p110\n\n"
        "Colors:\n"
        "Blue: 300K\n"
        "Green: 600K\n"
        "Red: 900K",
        xy=(0.02, 0.02),
        xycoords='axes fraction',
        bbox=dict(boxstyle="round,pad=0.5", facecolor="white", alpha=0.8),
        fontsize=12
    )
    
    # Save the figure
    plt.savefig(f"stress_strain_trial{trial}.png", dpi=300, bbox_inches='tight')
    
    # Show the plot
    plt.show()
```
