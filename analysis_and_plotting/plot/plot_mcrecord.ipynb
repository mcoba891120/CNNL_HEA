```python
import matplotlib.pyplot as plt

# 读取文件
file_path = 'monte_carlo/MC_5quinary/NiCoTiZrHf_110/edge/b100p100/v3_trial3_9792_300k_b100p100/mc_folder/MC_record.txt'
title = "Potential MC "
y_axis = "potetial"
x_axis = "step"
x = []
y = []

with open(file_path, 'r') as file:
    next(file)  # 跳过第一行
    for line in file:
        values = line.strip().split()
        if len(values) >= 2:
            try:
                x.append(float(values[0]))
                y.append(float(values[2]))
            except ValueError:
                print(f"Skipping invalid line: {line.strip()}")

# 创建图表
plt.figure(figsize=(10, 6))
plt.plot(x, y)

# 设置标签和标题
plt.xlabel(x_axis)
plt.ylabel(y_axis)
plt.title(title)

# 显示网格
plt.grid(True)

# 保存图表
plt.savefig('MC_record_plot_300.png')

# 显示图表
plt.show()

# 打印前几行数据以检查
print("First few data points:")
for i in range(min(5, len(x))):
    print(f"x: {x[i]}, y: {y[i]}")
```
