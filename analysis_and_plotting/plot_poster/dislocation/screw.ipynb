```python
import matplotlib.pyplot as plt
import numpy as np

# Data from the image
labels = ['b100_p100', 'b100_p110', 'b110_p100', 'b110_p110']
screw_values_300 = [0.097947473, 0.266228241, 1.227300957, 0.787496564]
screw_values_1273 = [0.117086039, 0.041138849, 1.674047742, 1.29155243]

# X-axis locations for the groups
x = np.arange(len(labels))

# Width of the bars
width = 0.35

# Plotting the bars
fig, ax = plt.subplots()
bars1 = ax.bar(x - width/2, screw_values_300, width, label='Screw 300 K')
bars2 = ax.bar(x + width/2, screw_values_1273, width, label='Screw 1273 K')

# Adding labels, title, and custom x-axis tick labels
# ax.set_xlabel('b_p_ Configurations')
ax.set_ylabel('Formation Energy (eV/Å)')
ax.set_title('Screw Dislocation Formation Energy')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.set_ylim(0, 2)  # Setting y-axis limits from 0 to 2
ax.legend()

# Display the plot
plt.xticks(rotation=45)
plt.show()
```
