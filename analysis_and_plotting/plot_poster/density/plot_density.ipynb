```python
import pandas as pd
import matplotlib.pyplot as plt

# 文件路径列表
file_paths = ['300_Record_npt.txt', '600_Record_npt.txt', '900_Record_npt.txt', 
              '1200_Record_npt.txt', '1500_Record_npt.txt']

# 列名
columns = ['Step', 'Temperature', 'Potential_Energy', 'Volume', 'Density', 'Pxx', 'Pyy', 'Pzz']

# 创建图形
plt.figure(figsize=(20, 10))

# 读取每个文件并绘制密度随时间步的变化
for file_path in file_paths:
    # 提取温度值作为图例
    temperature = file_path.split('_')[0]
    
    # 读取数据，跳过第一行注释
    data = pd.read_csv(file_path, sep=' ', header=None, names=columns, skiprows=1)
    
    # 计算平均密度
    average_density = data['Density'].mean()
    
    # 绘制密度随时间步变化图
    plt.plot(data['Step'], data['Density'], marker='o', linestyle='-', 
             label=f'{temperature}K (Avg Density: {average_density:.2f} g/cm$^3$)')
    
    # 添加横线，表示平均密度，但不在图例中显示
    plt.axhline(y=average_density, linestyle='--', color='gray')

# 设置图形参数
plt.xlabel('Step', fontsize='18')
plt.ylabel('Density (g/cm$^3$)', fontsize='18')
plt.title('Candidate 3  Density vs Step', fontsize='18')
plt.legend(title="Temperature and Avg Density", fontsize='16', title_fontsize='18')
plt.grid(True)
plt.show()
```
