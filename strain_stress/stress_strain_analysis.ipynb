### Draw the stress vs frequency and stress vs frequency curve of each slip system under this folder

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# dislocation_type = ['edge','screw']
# slip_systems = {'edge_b100_p100':46080,'edge_b100_p110':45696,'edge_b110_p110':46240,'edge_b111_p110':45696,
#                 'screw_b100_p100':46080,'screw_b100_p110':46240,'screw_b110_p110':45696,'screw_b111_p110':48960}
systems = {'mc300k':51840,'mc900k':51840}
mc_name = ['mc300k','mc900k']
atoms_types = ['Ni','Co','Ti','Zr']
```

```python

def plot_strain_vs_freqency():
    initial_dir = os.getcwd()
    mc_count = 0
    avg_strain = []
    for slip_system in list(systems.keys()):
        print(f"Processing slip system: {slip_system}")
        try:
            os.chdir(slip_system)
        except Exception as e:
            print(f"Failed to change directory to {slip_system}: {e}")
            continue
        
        plt.figure(figsize=(10, 6))
        sub_avg_strain = []
        data_present = False
        for atom_type in atoms_types:
            file_name = f"{atom_type}_strain.data"
            if not os.path.isfile(file_name):
                print(f"File {file_name} not found in {slip_system}")
                continue
            
            try:
                data = np.loadtxt(file_name)
                avg = np.mean(data[:,0])
                sub_avg_strain.append(avg)
                # Scaling the x-axis values
                data[:,1] = data[:,1] / systems[slip_system] *4 *100
                # Print data range for debugging
                print(f"{atom_type} data range - X: {data[:,0].min()} to {data[:,0].max()}, Y: {data[:,1].min()} to {data[:,1].max()}")
                
                plt.plot(data[:,0], data[:,1], label=atom_type, linestyle='--')
                data_present = True
                
            except Exception as e:
                print(f"Failed to process {file_name} in {slip_system}: {e}")
                continue
        avg_strain.append(sub_avg_strain)
        if not data_present:
            print(f"No data to plot for {slip_system}")
            plt.close()
            os.chdir(initial_dir)
            continue

        # Adjust axis limits based on data
        plt.xlim(0, 0.15)
        plt.ylim(0, 7.5)
        plt_title = slip_system+'_strain'
        plt.legend()
        plt.ylabel('Frequency %')
        plt.xlabel('Strain')
        
        plt.title(mc_name[mc_count]+'_strain')
        
        plot_file_name = slip_system +'_strain' + '.jpg'
        plt.savefig(plot_file_name, dpi=300, bbox_inches='tight')
        print(f"Saved plot as {plot_file_name}")
        plt.show()
        plt.close()
        mc_count += 1
        
        os.chdir(initial_dir)
    print(avg_strain)
```
