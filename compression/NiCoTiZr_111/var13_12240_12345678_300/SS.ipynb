```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Load the data
file_path = 'SS_curve_var13_12440_12345678.txt'
data = pd.read_csv(file_path, delim_whitespace=True)

# Plot strain vs stressx
plt.figure(figsize=(10, 6))
plt.plot(data['strain'], data['stressx'], label='StressX vs Strain', marker='o')

# Calculate and plot the slope
slope, intercept, r_value, p_value, std_err = linregress(data['strain'], data['stressx'])
plt.plot(data['strain'], intercept + slope * data['strain'], 'r', label=f'Fit line (slope={slope:.2f})')

plt.xlabel('Strain')
plt.ylabel('StressX')
plt.title('var13 Strain vs StressX with Fit Line (100)')
plt.legend()
plt.grid(True)
plt.show()
```
