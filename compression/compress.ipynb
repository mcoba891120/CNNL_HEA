```python
import matplotlib.pyplot as plt
import numpy as np

ss_100 = np.loadtxt('SS_curve_var1_13500_300k.txt',skiprows=1)
ss_110 = np.loadtxt('SS_curve_var1_13500_600k.txt',skiprows=1)

# ss_110 = np.loadtxt('SS_curve110.txt',skiprows=1)
# ss_111 = np.loadtxt('SS_curve111.txt',skiprows=1)

```

## 100,110,111 stress strain

```python
plt.figure()
plt.plot(ss_100[:,0], ss_100[:,1], label='var1_13500_300k')
plt.plot(ss_110[:,0], ss_110[:,1], label='var1_13500_600k')
# plt.plot(ss_111[:,0], ss_111[:,1], label='111')
slope_var1_13500_300k = np.polyfit(ss_100[:,0], ss_100[:,1], 1)[0]
slope_var1_13500_600k = np.polyfit(ss_110[:,0], ss_110[:,1], 1)[0]
# slope_111 = np.polyfit(ss_111[:,0], ss_111[:,1], 1)[0]
# print('slope 100:', slope_100, 'slope 110:', slope_110, 'slope 111:', slope_111)
print('slope 100:', slope_var1_13500_300k)
print('slope 110:', slope_var1_13500_600k)
plt.legend()
plt.title('SS curve')
plt.show()
```
