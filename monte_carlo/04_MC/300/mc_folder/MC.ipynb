```python
import matplotlib.pyplot as plt

def plot_step_vs_potential(file_path):
    # 初始化儲存 step 和 potential 的清單
    steps = []
    potentials = []

    # 讀取檔案
    with open(file_path, 'r') as file:
        lines = file.readlines()
        for line in lines:
            # 解析每一行並將數據加入清單，只取前兩個數字
            values = line.split()
            if len(values) >= 2:
                step, potential = map(float, values[:2])
                steps.append(step)
                potentials.append(potential)
    
    # 繪圖
    plt.figure(figsize=(12, 6))
    plt.plot(steps, potentials, marker='o', linestyle='-', color='b', markersize=0.001)
    plt.xlabel('Step')
    plt.ylabel('Potential')
    plt.title('Step vs Potential')
    plt.grid(True)
    plt.show()

# 使用範例檔案路徑
file_path = 'MC_record.txt'
plot_step_vs_potential(file_path)
```
